Оригинальный прототип

```python
# 1. Устанавливаем нужные библиотеки
!pip install transformers torch gTTS accelerate -q

import torch
from transformers import pipeline
from gtts import gTTS

# 2. Создаем тестовое аудио из текста (со смешанными словами)
text_for_audio = "Привет, мы тестируем распознавание речи и используем слово machine learning в нашем проекте."
tts = gTTS(text=text_for_audio, lang='ru')
audio_file = "test_audio.mp3"
tts.save(audio_file)
print(f"Мы создали тестовое аудио: {audio_file}")

# 3. Загружаем модель для распознавания речи
# Используем whisper-small: она лучше понимает контекст и смешанные языки, чем base
device = "cuda" if torch.cuda.is_available() else "cpu"
asr_pipeline = pipeline(
    "automatic-speech-recognition",
    model="openai/whisper-small",
    device=device
)
print("Модель загружена, мы готовы к распознаванию.")

# 4. Отправляем аудио в модель и получаем текст
result = asr_pipeline(audio_file, return_timestamps=True)
recognized_text = result["text"]

print(f"Распознанный текст: {recognized_text}")
```

Тест с распознавванием записи всвтречи от 2026-09-30

```python
# 1. Устанавливаем нужные библиотеки
!pip install transformers torch accelerate -q

import torch
from transformers import pipeline
from google.colab import files

# 2. Загружаем аудиофайл с компьютера
print("Загрузите ваш аудиофайл через появившееся окно...")
uploaded = files.upload()

file_path = list(uploaded.keys())[0]
print(f"Мы успешно загрузили файл: {file_path}")

# 3. Загружаем модель для распознавания речи
device = "cuda" if torch.cuda.is_available() else "cpu"
asr_pipeline = pipeline(
    "automatic-speech-recognition",
    model="openai/whisper-base",
    device=device
)
print("Модель загружена, мы начинаем транскрибацию.")

# 4. Отправляем загруженное аудио в модель и получаем текст
# Мы добавили return_timestamps=True, чтобы модель могла работать с длинными файлами
result = asr_pipeline(file_path, return_timestamps=True)
recognized_text = result["text"]

print(f"Распознанный текст: {recognized_text}")
```
